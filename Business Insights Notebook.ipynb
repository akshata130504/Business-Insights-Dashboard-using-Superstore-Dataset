# Business Insights Dashboard using Superstore Dataset

## 📖 Introduction
This notebook analyzes the Superstore dataset to extract meaningful business insights.
The goal is to transform raw data into actionable insights that support decision-making.

## 🎯 Objectives
- Analyze sales and profit trends
- Identify loss-making areas
- Understand category and regional performance

```python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Style
sns.set(style="whitegrid")

# Load dataset
df = pd.read_csv(r'C:\Users\Rathod\Downloads\Sample - Superstore.csv', encoding='latin1')

# Preview data
df.head()
```

## 🔍 Data Understanding
Understanding dataset structure, columns, and data types.

```python
# Shape of dataset
df.shape
```

```python
# Dataset info
df.info()
```

```python
# Column names
df.columns
```

##  Data Cleaning
Ensuring data quality by handling missing values and duplicates.

```python
# Check null values
df.isnull().sum()
```

```python
# Remove duplicates
df.drop_duplicates(inplace=True)
```

```python
# Verify shape
df.shape
```

## 📊 Basic Data Analysis
Performing aggregations to understand key metrics.

```python
# Total Sales
print("Total Sales:", df['Sales'].sum())

# Total Profit
print("Total Profit:", df['Profit'].sum())

# Sales by Category
sales_category = df.groupby('Category')['Sales'].sum()
sales_category

# Profit by Region
profit_region = df.groupby('Region')['Profit'].sum()
profit_region

# Discount vs Profit correlation
df[['Discount', 'Profit']].corr()
```

## 📈 Data Visualization
Visualizing trends and relationships in the dataset.

```python
plt.figure()

sales_category.plot(
    kind='bar',
    color=['#FF6F61', '#6B5B95', '#88B04B']  # pink, purple, green
)

plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales")
plt.xticks(rotation=0)
plt.show()
```

## Insight:
Technology category shows the highest sales, indicating it is the main contributor to overall revenue compared to Furniture and Office Supplies.

```python
plt.figure()

profit_region.plot(
    kind='bar',
    color=['#FFA500', '#00BFFF', '#32CD32', '#FF69B4']  # orange, blue, green, pink
)

plt.title("Profit by Region")
plt.xlabel("Region")
plt.ylabel("Profit")
plt.xticks(rotation=0)
plt.show()
```

## Insight:
The West region generates the highest profit, while other regions show comparatively lower profitability, suggesting regional performance differences.

```python
plt.figure()

plt.scatter(
    df['Discount'],
    df['Profit'],
    color='#FF1493',  # bright pink
    alpha=0.6
)

plt.title("Discount vs Profit")
plt.xlabel("Discount")
plt.ylabel("Profit")
plt.show()
```

## Insight
Higher discounts tend to reduce profit, and in some cases lead to losses, indicating that excessive discounting negatively impacts profitability.

```python
plt.figure(figsize=(10,5))

# Convert to datetime
df['Order Date'] = pd.to_datetime(df['Order Date'])

# Create Year-Month
df['YearMonth'] = df['Order Date'].dt.to_period('M')

# Group
sales_time = df.groupby('YearMonth')['Sales'].sum()

# Convert index
sales_time.index = sales_time.index.astype(str)

# Plot
plt.plot(sales_time, color='#1E90FF', marker='o')

# Reduce x-axis clutter (show every 3rd label)
plt.xticks(ticks=range(0, len(sales_time), 3), rotation=45)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.tight_layout()
plt.show()
```

## Insight:

Monthly sales show fluctuations over time with certain peak periods, indicating seasonal variations in customer demand. This suggests that the business experiences higher sales during specific months, which can be targeted for marketing and inventory planning.

```python
plt.figure()

top_products = df.groupby('Product Name')['Sales'].sum().sort_values(ascending=False).head(10)

top_products.plot(
    kind='barh',
    color='#32CD32'  # green
)

plt.title("Top 10 Products by Sales")
plt.xlabel("Sales")
plt.gca().invert_yaxis()  # highest on top
plt.show()
```

## Insight:

A small number of products contribute significantly to total sales, indicating that the business relies heavily on top-performing products. Focusing on these products can help maintain strong revenue growth.

```python
plt.figure(figsize=(8,5))

category_region = df.pivot_table(
    values='Sales',
    index='Category',
    columns='Region',
    aggfunc='sum'
)

category_region.plot(
    kind='bar',
    stacked=True,
    colormap='Set2'
)

plt.title("Sales by Category & Region")
plt.ylabel("Sales")
plt.xticks(rotation=0)
plt.show()
```

## Insight:

The chart shows that sales for each category are not evenly distributed across regions, indicating regional differences in product demand. Identifying high-performing region-category combinations can help improve targeted marketing and sales strategies.

```python
plt.figure(figsize=(6,6))

region_sales = df.groupby('Region')['Sales'].sum()

colors = ['#FF6F61', '#6B5B95', '#88B04B', '#FFA500']

plt.pie(
    region_sales,
    labels=region_sales.index,
    autopct='%1.1f%%',
    colors=colors,
    startangle=90
)

# Donut effect
centre_circle = plt.Circle((0,0), 0.70, fc='white')
plt.gca().add_artist(centre_circle)

plt.title("Sales by Region")
plt.show()
```

## Insight:

Sales are not evenly distributed across regions, with some regions contributing a larger share of total revenue. This indicates that certain regions perform better and play a key role in overall business growth.
